#Khai báo thư viện


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#Các thống kê

```python
path_to_dataset = '/content/drive/MyDrive/Python for ML/dataset/cleaned_all_phones.csv'
```

```python
df = pd.read_csv(path_to_dataset)

df.head()
```

```python
# Thống kê số lượng cột và tên cột
print("Số lượng cột:", len(df.columns))
print("Tên các cột:", df.columns.tolist())

# Số lượng hàng (ngoại trừ hàng tiêu đề cột)
num_rows = df.shape[0]
print("\nSố lượng hàng:", num_rows)
```

```python
# Kiểm tra kiểu dữ liệu của từng cột
print("\nKiểu dữ liệu của từng cột:")
print(df.dtypes)

# Số lượng NaN, NULL trong mỗi cột
nan_counts = df.isna().sum()
print("\nSố lượng NaN trong mỗi cột:")
print(nan_counts)
```

```python
# Các cột là biến phân loại (categorical)
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

# Các cột là biến số (numerical)
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()

print("\nCác cột là biến phân loại:", categorical_cols)
print("Các cột là biến số:", numerical_cols)
```

```python
# Cài đặt để vẽ biểu đồ
plt.figure(figsize=(12, 6))

# Biểu đồ số lượng của mỗi brand
plt.subplot(1, 2, 1)
sns.countplot(x='brand', data=df)
plt.title('Số lượng mỗi brand')

# Biểu đồ phân phối của cột 'price(USD)'
plt.subplot(1, 2, 2)
sns.histplot(df['price(USD)'], kde=True, bins=30)
plt.title('Vùng phân bố giá trị của price(USD)')

plt.tight_layout()
plt.show()
```
